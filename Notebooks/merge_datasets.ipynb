## Split DF into Relationship Types

There are four relationship types:
* 1: The boundaries of a single CoC and a single county are coterminous.
* 2: A CoC's boundary fully encompass multiple counties.
* 3: A single county fully contains multiple CoCs.
* 4: A CoC is partially located in multiple counties and the counties in which it is located also contain parts of multiple CoCs.

## Create Network of Geographies

```python
matches
```

# Create Groups from The Edge List

```python
import networkx as nx

# Create a graph from the DataFrame edgelist
G = nx.from_pandas_edgelist(matches, source='coc_number', target='county_fips')

connected_graphs = list(nx.connected_components(G))

len(connected_graphs)
```

We have 347 Groups containing all of our CoC/County elements.

# Create a LookUp Map

We will create a dictionary so we can lookup each FIPS number or Coc number and see which group it is in.

```python
# create a dictionary
element_to_group_map = {}

# Use enumerate() to get both the index (0, 1, 2...) and the list of FIPS
for group_index, elements in enumerate(connected_graphs):
    for element in elements:
        # We add 1 to the index to get group numbers 1-347
        # instead of 0-346.
        element_to_group_map[element] = group_index + 1
```

# Apply Map to Data

```python
acs_data['Group_Number'] = acs_data['GEOID'].map(element_to_group_map)
hic_counts['Group_Number'] = hic_counts['COC_Number'].map(element_to_group_map)
pit_counts['Group_Number'] = pit_counts['CoC Number'].map(element_to_group_map)
print(acs_data['Group_Number'].isna().sum())
print(len(acs_data))
print(hic_counts['Group_Number'].isna().sum())
print(len(hic_counts))
print(pit_counts['Group_Number'].isna().sum())
print(len(pit_counts))
```

# Aggregate the Data Per Group

```python
acs_data = acs_data.rename(columns={'Year':'year'})
acs_data.columns
```

```python
agg_acs_data = acs_data.groupby(['Group_Number','year']).agg(
    # New_Column_Name = ('Original_Column_Name', 'Function')
    Total_Population   = ('TotalPopulation', 'sum'),
    Central_Tendency_Rent = ('MedianRent', 'mean'),
    Central_Tendency_Income = ('MedianIncome', 'mean'),
    Poverty_Numerator = ('PovertyNumerator', 'sum'),
    Poverty_Denominator = ('PovertyDenominator', 'sum'),
    Vacancy_Numerator = ('VacancyNumerator', 'sum'),
    Vacancy_Denominator = ('VacancyDenominator', 'sum'),
    Renter_Occupied_Numerator = ('RenterOccupiedNumerator', 'sum'),
    Renter_Occupied_Denominator = ('RenterOccupiedDenominator', 'sum'),
    Rent_Burdened_Numerator = ('RentBurdenedNumerator', 'sum'),
    Rent_Burdened_Denominator = ('RentBurdenedDenominator', 'sum'),
    Unemployment_Numerator = ('UnemploymentNumerator', 'sum'),
    Unemployment_Denominator = ('UnemploymentDenominator', 'sum')
)

print(agg_acs_data.isna().sum())
print(len(agg_acs_data))
```

```python
hic_counts.columns
```

```python
year_mask = (hic_counts['year'] >= 2009) & (hic_counts['year'] <= 2023)
agg_hic_count = hic_counts[year_mask].groupby(['Group_Number','year']).agg(
    # New_Column_Name = ('Original_Column_Name', 'Function')
    Beds = ('Total_Year_Round_Beds', 'sum'))

print(len(agg_hic_count))
print(agg_hic_count.isna().sum())
```

```python
pit_counts.columns
```

```python
year_mask = (pit_counts['year'] >= 2009) & (pit_counts['year'] <= 2023)
agg_pit_count = pit_counts[year_mask].groupby(['Group_Number','year']).agg(
    # New_Column_Name = ('Original_Column_Name', 'Function')
    Overall_Homeless = ('Overall Homeless', 'sum'),
    Unsheltered_Homeless = ('Unsheltered Homeless', 'sum'),
    Sheltered_Total_Homeless = ('Sheltered Total Homeless', 'sum'),
    Overall_Homeless_Individuals = ('Overall Homeless Individuals', 'sum'),
    Overall_Homeless_People_in_Families = ('Overall Homeless People in Families', 'sum')
)
print(len(agg_pit_count))
print(agg_pit_count.isna().sum())
```

```python
merged_data = pd.merge(left=agg_pit_count, right=agg_hic_count, on=['Group_Number','year'])
merged_data = pd.merge(left=merged_data, right=agg_acs_data, on=['Group_Number','year'])
merged_data = merged_data.reset_index()
```

```python

merged_data
```

```python
Final_Data_Set = merged_data[['Group_Number','year','Overall_Homeless', 'Unsheltered_Homeless', 'Sheltered_Total_Homeless', 'Overall_Homeless_Individuals', 'Overall_Homeless_People_in_Families', 'Beds', 'Total_Population', 'Central_Tendency_Rent', 'Central_Tendency_Income' ]].copy()
Final_Data_Set['Poverty_Rate'] = merged_data['Poverty_Numerator'] / merged_data['Poverty_Denominator']
Final_Data_Set['Vacancy_Rate'] = merged_data['Vacancy_Numerator'] / merged_data['Vacancy_Denominator']
Final_Data_Set['Renter_Occupied_Rate'] = merged_data['Renter_Occupied_Numerator'] / merged_data['Renter_Occupied_Denominator']
Final_Data_Set['Rent_Burdened_Rate'] = merged_data['Rent_Burdened_Numerator'] / merged_data['Rent_Burdened_Denominator']
Final_Data_Set['Unemployment_Rate'] = merged_data['Unemployment_Numerator'] / merged_data['Unemployment_Denominator']
Final_Data_Set
```

```python
Final_Data_Set.to_csv('../Data/Final_Data_Set.csv',index=False)
```
