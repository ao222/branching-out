```python
import pandas as pd
```

```python
dfs = pd.read_excel('../Data/raw/2007-2024-HIC-Counts-by-CoC.xlsx', sheet_name=None)
new_col_names = ["COC_Number", "Total_Year_Round_Beds"]
dfs2 = []
for name, sheets in dfs.items():
    if name.isdigit():
        year = int(name)
        temp = sheets.iloc[1:,:2].rename(columns=dict(zip(sheets.columns,new_col_names)))
        temp['year'] = year
        dfs2.append(temp)
```

```python
HIC_count = pd.concat(dfs2, ignore_index=True)
```

```python
HIC_count['year'].unique()
```

```python
HIC_count.to_csv("hic_counts.csv", index=False)
```
