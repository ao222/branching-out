```python
def remove_outliers_iqr(df, column_name):
    """
    Removes outliers from a DataFrame based on the IQR method for a specific column.

    The IQR method defines outliers as any value falling outside the range:
    [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR]



    Parameters:
    df (pd.DataFrame): The input DataFrame.
    column_name (str): The name of the numeric column to check for outliers.

    Returns:
    pd.DataFrame: A new DataFrame with the outlier rows removed.
    """
    # 1. Check if the column exists
    if column_name not in df.columns:
        print(f"Error: Column '{column_name}' not found in DataFrame.")
        return df

    # 2. Check if the column is numeric
    if not pd.api.types.is_numeric_dtype(df[column_name]):
        print(f"Warning: Column '{column_name}' is not numeric. IQR method not applicable.")
        return df

    # 3. Calculate Q1 (25th percentile) and Q3 (75th percentile)
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)

    # 4. Calculate the Interquartile Range (IQR)
    IQR = Q3 - Q1

    # 5. Define the outlier bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # 6. Create a mask for values *within* the bounds (inliers)
    # This automatically handles NaN values (they become False in the mask)
    mask = (df[column_name] >= lower_bound) & (df[column_name] <= upper_bound)

    # 7. Apply the mask to the DataFrame
    df_cleaned = df[mask]

    # 8. Report the results
    rows_removed = len(df) - len(df_cleaned)
    if rows_removed > 0:
        print(f"Removed {rows_removed} outlier(s) from '{column_name}' column.")
    else:
        print(f"No outliers found in '{column_name}'.")

    return df_cleaned
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from pyod.models.iforest import IForest

def remove_outliers_pyod(df, numeric_columns, contamination=0.05):
    """
    Removes outliers from a DataFrame using the PyOD Isolation Forest (IForest) model.

    This function considers all specified numeric columns at once to find
    multivariate outliers.

    Parameters:
    df (pd.DataFrame): The input DataFrame.
    numeric_columns (list): A list of column names to use for detection.
                            (This should be all 10 of your numeric columns).
    contamination (float): The expected proportion of outliers in the data,
                           between 0 and 0.5. Default is 0.05 (5%).

    Returns:
    pd.DataFrame: A new DataFrame with the outlier rows removed.
    """
    print("--- Starting outlier detection ---")

    # Create a copy to avoid modifying the original DataFrame
    df_copy = df.copy()

    # 1. Preprocessing: Select only the numeric columns for the model
    # The model requires data to be clean (no NaN)
    df_numeric = df_copy[numeric_columns].dropna()

    # Store the original indices so we can filter the main DataFrame
    original_indices = df_numeric.index

    if df_numeric.empty:
        print("Error: DataFrame is empty after dropping NaN values.")
        return df

    # 2. Preprocessing: Scale the data
    # While IForest is not sensitive to scaling, it's good practice
    # for many other PyOD models (like KNN or LOF).
    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(df_numeric)

    # 3. Initialize and Train the Model
    # 'contamination' is the most important parameter.
    # It's your assumption of what % of your data is anomalous.
    print(f"Initializing IForest with {contamination*100}% contamination...")
    clf = IForest(contamination=contamination,
                  random_state=42,
                  n_estimators=100)

    # Train the model and get predictions (0 = inlier, 1 = outlier)
    # fit_predict() is a handy shortcut for .fit().predict()
    y_pred = clf.fit_predict(data_scaled)

    # 4. Add predictions to the numeric DataFrame to see results
    df_numeric['is_outlier'] = y_pred

    # Count the outliers found
    outliers_found = df_numeric['is_outlier'].sum()
    print(f"Total outliers found: {outliers_found}")

    # 5. Filter the *Original* DataFrame
    # Get the indices of the *inliers* (where is_outlier == 0)
    inlier_indices = df_numeric[df_numeric['is_outlier'] == 0].index

    # Use .loc to select only the inlier rows from the original DataFrame
    df_cleaned = df_copy.loc[inlier_indices]

    print(f"Original rows: {len(df)}. Cleaned rows: {len(df_cleaned)}.")
    return df_cleaned
```

# Light Gradient Boosting Machine

## Library Imports

```python
import lightgbm as lgb
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

## Define Dataset and Parameters

```python
# Create LightGBM Dataset objects
lgb_train = lgb.Dataset(X_train, y_train)
lgb_eval = lgb.Dataset(X_test, y_test, reference=lgb_train)
```

```python
params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 31,  # Max number of leaves in one tree
    'learning_rate': 0.05,
    'feature_fraction': 0.9, # Fraction of features to be randomly selected
    'verbose': -1 # Suppress logging
}
```

## Train the Model

```python
num_iterations = 100

gbm = lgb.train(
    params,
    lgb_train,
    num_iterations,
    valid_sets=lgb_eval
)
```

```python
# Make predictions on the test set
y_pred2 = gbm.predict(X_test, num_iteration=gbm.best_iteration)

mse = mean_squared_error(y_test, y_pred2)

r2 = r2_score(y_test, y_pred2)

print(mse)
print(r2)
```
